# stratification by mets at enrollment

```python
#cluster ptIDs (keys) from metEnrollmentDict by similar values in list
metlocations = [1,2,3,4,5,6]
metSort = {} #keys are met locations, values are patient IDs with that met location at enrollment. may have other mets
metSortSingle = {} #like above, but only has the one met location

for location in metlocations: 
    metLocationTemp = [] #list of patient IDs that have met at enrollment at a given location
    metSingleTemp = [] #list of patient IDs where this location is the only met location at enrollment
    for ptID in metEnrollmentDict.keys():
        if location in metEnrollmentDict[ptID]:
           metLocationTemp.append(ptID)
           if len(metEnrollmentDict[ptID]) == 1: 
              metSingleTemp.append(ptID)
    metSort[location] = metLocationTemp
    metSortSingle[location]= metSingleTemp
```

# Exporting data to spreadsheets

```python
folder_path_processed = "012_Processed_Data"
folder_out_header = os.path.join(computer_path_header, folder_path2, folder_path_processed)
```

```python
#to excel function
def outputToExcel(df_data, fileName_header, fileName_Common, fileName_suffix,
                  parent_dir, folderName, sheetName):
    fileName = fileName_header + fileName_Common + fileName_suffix
    outfile_extension = '.xlsx'
    outfile_boxplts = fileName + outfile_extension
    path_out= os.path.join(parent_dir, folderName, outfile_boxplts)

    if os.path.exists(path_out):
            #if old sheet
            with pd.ExcelWriter(path_out, mode = 'a', if_sheet_exists = 'overlay') as writer:
                    df_data.to_excel(writer, sheet_name = sheetName, index = True)
    else: 
            #new sheet
            with pd.ExcelWriter(path_out) as writer:
                    df_data.to_excel(writer, sheet_name = sheetName, index = True) #if new sheet
```

```python
# tumor2ptdict
#each column is a different primary tumor type
outputToExcel(pd.DataFrame.from_dict(tumor2ptdict, orient='index' ).T, fileName_header = "primaryTumorStrat", fileName_Common = "", fileName_suffix = "", parent_dir = folder_out_header, folderName = "", sheetName = "primary tumor")
```

```python
# mets2Dict
# the column is the met type, the sheet is the primary tumor type
for j in mets2Dict.keys():
    outputToExcel(pd.DataFrame.from_dict(mets2Dict[j], orient='index' ).T, fileName_header = "primaryTumorStrat_mets", fileName_Common = "", fileName_suffix = "", parent_dir = folder_out_header, folderName = "", sheetName = str(j))
```
